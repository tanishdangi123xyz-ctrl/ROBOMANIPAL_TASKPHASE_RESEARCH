```python
import math
import pandas as pd
```

```python
def KNN(x, y, X, Y, k, classes):
    distances = []
    for i in range(len(X)):
        distance = math.sqrt(((X[i] - x) ** 2) + ((Y[i] - y) ** 2))
        distances.append((distance, classes[i]))

    distances.sort(key = lambda x: x[0])
    kn_classes = []
    for i in range(k):
        kn_classes.append(distances[i][1])

    votes = {}
    for i in range(len(kn_classes)):
        votes[kn_classes[i]] = votes.get(kn_classes[i], 0) + 1

    maj_class = max(votes, key=votes.get)
    return maj_class
```

```python
X = [1.5, 2.0, 1.8, 3.0, 7.5, 8.2, 9.0, 8.5]
Y = [2.1, 1.5, 2.8, 3.2, 8.0, 7.2, 8.5, 9.1]
classes = [0, 0, 0, 0, 1, 1, 1, 1]
```

```python
cls = KNN(2.6, 3.2, X, Y, 3, classes)
print(cls)
```
